## 3. Document Processing and Tokenization
Prepare unstructured data for the context window.


````python
from langchain_text_splitters import CharacterTextSplitter

# Example sample text (could be Toronto Open Data descriptions)
text = "The City of Toronto provides open data on traffic, public services, and urban operations."

text_splitter = CharacterTextSplitter(chunk_size=1000, chunk_overlap=0)
texts = text_splitter.split_text(text)
print(f"Created {len(texts)} chunks")
````
